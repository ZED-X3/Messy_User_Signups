```python
import re
import pandas as pd
import numpy as np

user_data = pd.read_csv('messy_user_signups.csv')

print('Original rows:', len(user_data))
print('Unique user IDs:', user_data['user_id'].nunique())
print('Duplicate user IDs:', user_data['user_id'].duplicated().sum())
print('Missing emails:', user_data['email'].isna().sum())
```

```python
def parse_date(s):
    if pd.isna(s):
        return pd.NaT

    s = str(s).strip().replace('/', '-').replace('.', '-')

    fmt = '%Y-%m-%d' if re.match(r'^\d{4}-', s) else '%d-%m-%Y'

    try:
        return pd.to_datetime(s, format=fmt)
    except ValueError:
        return pd.NaT

user_data['signup_date'] = user_data['signup_date'].apply(parse_date)
```

```python
user_data = user_data.drop(index= [30, 31, 6, 37, 11, 33, 34, 22, 27, 39])
```

```python
user_data['country'] = (
    user_data['country']
    .str.strip()
    .str.title()
    .replace({
        'Korea': 'South Korea',
        'Usa': 'United States'
    })
)
user_data['name'] = user_data['name'].str.strip()
```

```python
print('Cleaned rows:', len(user_data))
print('Unique user IDs:', user_data['user_id'].nunique())
print('Duplicate user IDs:', user_data['user_id'].duplicated().sum())
print('Missing emails:', user_data['email'].isna().sum())
print('Missing dates:', user_data['signup_date'].isna().sum())
```
